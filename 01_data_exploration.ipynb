```python
import re

# 1. Define our extraction functions using Regex
def extract_btc_address(text):
    # Standard regex for legacy (P2PKH) Bitcoin addresses starting with 1
    # Real implementations would include SegWit (bc1) and other chains
    btc_pattern = r'\b[13][a-km-zA-HJ-NP-Z1-9]{25,34}\b'
    matches = re.findall(btc_pattern, text)
    return matches[0] if matches else None

def extract_pgp_key(text):
    # Regex to find 16-character hexadecimal PGP fingerprints 
    # (allowing optional spaces for readability)
    pgp_pattern = r'\b([A-Fa-f0-9]{4}\s?[A-Fa-f0-9]{4}\s?[A-Fa-f0-9]{4}\s?[A-Fa-f0-9]{4})\b'
    matches = re.findall(pgp_pattern, text)
    # Remove spaces for consistency if found
    return matches[0].replace(' ', '') if matches else None

# 2. Apply the functions to create new columns in our dataframe
df['extracted_btc'] = df['content'].apply(extract_btc_address)
df['extracted_pgp'] = df['content'].apply(extract_pgp_key)

# 3. View the enriched dataset
df[['author', 'forum', 'extracted_btc', 'extracted_pgp']]
```
